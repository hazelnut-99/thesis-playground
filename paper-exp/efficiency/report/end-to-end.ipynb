```python
def filter_df(df, filters):
    filtered_df = df.copy()
    
    for column, value in filters.items():
        if column in df.columns:
            filtered_df = filtered_df[filtered_df[column] == value]
        else:
            print(f"Warning: Column '{column}' not found in DataFrame")
    
    return filtered_df
```

```python
filtered_data = filter_df(df, {'allocator': 'LRU', 'trace_name': 'twitter_cluster49', 'wsr': 0.01})
filtered_data.sort_values(by=['miss_ratio'])[['rebalance_strategy', 'miss_ratio', 'rebalanced_slabs', 'uuid']]
```
